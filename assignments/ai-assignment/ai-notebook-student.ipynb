### Background
People often optimize for long-term goals when it comes to collaboration (Liu et al., 2026). In other words, a good mentor or companion doesn't just answer questions, but tracks progress and prioritizes a person's growth over immediate rewards. Conversely, current AI systems are optimized for providing instant and complete responses.


<div style="background-color: #EAF3FA; padding: 16px 20px; border-radius: 8px; margin-bottom: 20px;">

### 🔎 Research question

What are the consequences of AI systems prioritizing immediate assistance over long-term learning and independent performance?

</div>


### Experimental design
We'll be focusing on their first experiment. For this experiment, participants completed a fraction-solving task. Participants were randomly assigned to either an __AI__ condition, where they could use AI for assistance, or a __control__ condition, where they could not. They were tasked with solving a series of 12 fraction problems. To test independent problem solving, the AI assistant was removed without warning, and participants were tasked with solving three additional fraction problems. 

The __main measure__ for this experiment is the comparison of fraction __solve-rate__ and __skip-rate__ between conditions during the __test__ phase.
